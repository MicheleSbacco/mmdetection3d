```python
from json_handler import JSONHandler
handler = JSONHandler("./losses_log.json")
dictionary_list = handler.read_json_file()



# Remove the tail
i = -1
while dictionary_list[i]['type'] == "training":
    i -= 1
print(i)
dictionary_list_without_tail = dictionary_list[:i+1]
print(dictionary_list_without_tail[-1])
print(dictionary_list_without_tail[-2])
```

```python
# Lista per memorizzare i risultati finali
results = []

# Variabili per calcolare la media della "total_loss"
current_group = []
total_loss_sum = 0.0
training_count = 0

# Scorri tutti i dizionari nel file JSON
for entry in dictionary_list_without_tail:
    if entry["type"] == "training":
        # Aggiungi il valore di "total_loss" al totale e incrementa il contatore
        total_loss_sum += entry["total_loss"]
        training_count += 1
    elif entry["type"] == "validation":
        # Se è stato registrato almeno un dizionario di "training" contiguo
        if training_count > 0:
            # Calcola la media del gruppo corrente e aggiungilo ai risultati
            average_loss = total_loss_sum / training_count
            results.append({"type": "training", "total_loss": average_loss})
            # Resetta le variabili per il nuovo gruppo
            total_loss_sum = 0.0
            training_count = 0
        results.append(entry)

# Output dei risultati
for element in results:
    print(element)

import json
# Salva i risultati in un nuovo file JSON (opzionale)
with open('grouped_losses.json', 'w') as outfile:
    json.dump(results, outfile, indent=4)
```
